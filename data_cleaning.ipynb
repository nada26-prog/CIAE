```python
# Step 1: Load Your Data
data = pd.read_csv("mrna_data_cbioportal")
```

```python
data.drop("Entrez_Gene_Id", axis = 1, inplace = True)
```

```python
# Set index using the correct column name (identified from the printed columns)
index_col = "Hugo_Symbol"
data = data.set_index(index_col).T
```

```python
import numpy as np
import pandas as pd

# log2(x + 1) transformation
df_log2 = np.log2(data + 1)
```

# Preprocesssing of mRNA seq

```python
bpca_imputer = IterativeSVD(rank=50)
df = pd.DataFrame(bpca_imputer.fit_transform(df_log2), index=df_log2.index, columns=df_log2.columns)

print("BPCA Imputation Done!")
```

```python
from sklearn.preprocessing import StandardScaler
import numpy as np

# Create a scaler
scaler = StandardScaler()

# Normalize the data
normalized_data = scaler.fit_transform(df)

print(normalized_data)
```

```python
# Create a DataFrame for the reduced data with original column names
norm_df = pd.DataFrame(data=normalized_data, index=df.index, columns=df.columns)
```

```python
# 1. Filter low-expression genes
threshold = 0.01  # minimum expression threshold

# Median expression per gene
median = norm_df.median(axis=0)

# Proportion of samples where expression > threshold
prop_expr = (norm_df > threshold).sum(axis=0) / norm_df.shape[0]

# Keep genes that are not too flat or too rare
keep = (median > threshold) & (prop_expr >= 0.2)

# Filter the dataframe
Xf = norm_df.loc[:, keep]

print(f"Kept {Xf.shape[1]} genes out of {norm_df.shape[1]} ({100 * Xf.shape[1] / norm_df.shape[1]:.2f}%)")
```

```python
from scipy.stats import mstats
import pandas as pd

# norm_df = your z-scored DataFrame (samples x genes)
# Winsorize each column (top and bottom 1%)
X_winsor = Xf.apply(lambda x: mstats.winsorize(x, limits=[0.1, 0.1]))
X_winsor = X_winsor.astype(float)

# Now X_winsor is ready for AE input
print(X_winsor.min().min(), X_winsor.max().max())
```

```python
# Create a DataFrame for the reduced data with original column names
norm_df = pd.DataFrame(data=X_winsor, index=Xf.index, columns=Xf.columns)
```
